```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Sample data: Input parameters (features) and corresponding outputs (quality and cost)
# You should replace this with your actual dataset
# For simplicity, I'm generating random data here
np.random.seed(0)
num_samples = 1000
light = np.random.uniform(low=1000, high=5000, size=num_samples)
nutrient_solution = np.random.uniform(low=0.1, high=1.0, size=num_samples)
temperature = np.random.uniform(low=15, high=30, size=num_samples)
co2_level = np.random.uniform(low=300, high=1000, size=num_samples)
microbiome = np.random.uniform(low=0, high=1, size=num_samples)
pest_management = np.random.randint(0, 2, size=num_samples)  # 0 or 1 representing absence or presence of pest management

# Output: Quality and cost
# For simplicity, I'm generating these as linear combinations of input parameters
quality = 0.5 * light + 0.03 * nutrient_solution + 0.02 * temperature + 0.1 * co2_level + 0.01 * microbiome - 0.05 * pest_management
cost = 0.4 * light + 0.2 * nutrient_solution + 0.3 * temperature + 0.15 * co2_level + 0.05 * microbiome + 0.1 * pest_management

# Reshape features into a 2D array
X = np.column_stack((light, nutrient_solution, temperature, co2_level, microbiome, pest_management))

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, np.column_stack((quality, cost)), test_size=0.2, random_state=42)

# Train linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Evaluate the model
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
# Save the model
import joblib
joblib.dump(model, 'vertical_farming_model.pkl')
```

```python
import numpy as np
import joblib
# Load the trained model
model = joblib.load('vertical_farming_model.pkl')
# Example input parameters
light = 1000
nutrient_solution = 0.5
temperature = 25
co2_level = 700
microbiome = 0.8
pest_management = 1  # Presence of pest management
# Make predictions
input_data = np.array([[light, nutrient_solution, temperature, co2_level, microbiome, pest_management]])
quality, cost = model.predict(input_data)[0]

print("Predicted Quality:", quality)
print("Predicted Cost:", cost)
```
